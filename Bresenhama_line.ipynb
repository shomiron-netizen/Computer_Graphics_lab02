```python
import matplotlib.pyplot as plt
```

```python
def bresenham_line(x1, y1, x2, y2):

    pixels = []

    print("\n--- Bresenham's Line Algorithm ---")
    print("Starting Point:", (x1, y1))
    print("Ending Point:", (x2, y2))

    dx = abs(x2 - x1)
    dy = abs(y2 - y1)

    sx = 1 if x1 < x2 else -1
    sy = 1 if y1 < y2 else -1

    error = dx - dy

    while True:

        pixels.append((x1, y1))

        print(
            "Pixel ->",
            (x1, y1),
            "Decision/Error =", error
        )

        if x1 == x2 and y1 == y2:
            break

        e2 = 2 * error

        if e2 > -dy:
            error = error - dy
            x1 = x1 + sx

        if e2 < dx:
            error = error + dx
            y1 = y1 + sy

    return pixels
```

```python
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))
```

```python
pixels = bresenham_line(x1, y1, x2, y2)

print("\nGenerated Pixels:")
print(pixels)
```

```python
x_pixels = [p[0] for p in pixels]
y_pixels = [p[1] for p in pixels]

plt.figure(figsize=(8, 6))

plt.plot(
    [x1, x2],
    [y1, y2],
    label="Ideal Mathematical Line"
)

plt.scatter(
    x_pixels,
    y_pixels,
    s=100,
    marker="s",
    color="red",
    label="Bresenham Pixels"
)

plt.scatter(
    [x1, x2],
    [y1, y2],
    s=150,
    marker="o",
    color="green",
    label="End Points"
)

for x, y in pixels:
    plt.text(
        x + 0.1,
        y + 0.1,
        f"({x},{y})",
        fontsize=9
    )

plt.xticks(
    range(min(x1, x2) - 1, max(x1, x2) + 2)
)

plt.yticks(
    range(min(y1, y2) - 1, max(y1, y2) + 2)
)

plt.grid(True)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.title("Bresenham's Line Algorithm")

plt.legend()

plt.show()
```
