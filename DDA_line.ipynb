```python
import matplotlib.pyplot as plt
```

```python
def dda_line(x1, y1, x2, y2):

    pixels = []

    print("\n--- DDA Line Algorithm ---")
    print("Starting Point:", (x1, y1))
    print("Ending Point:", (x2, y2))

    dx = x2 - x1
    dy = y2 - y1

    steps = max(abs(dx), abs(dy))

    x_increment = dx / steps
    y_increment = dy / steps

    print("dx =", dx)
    print("dy =", dy)
    print("Steps =", steps)
    print("X =", x_increment)
    print("Y =", y_increment)

    x = x1
    y = y1

    for i in range(steps + 1):

        pixel_x = round(x)
        pixel_y = round(y)

        pixels.append((pixel_x, pixel_y))

        print(
            "x =", x,
            "y =", y,
            "Pixel ->", (pixel_x, pixel_y)
        )

        x = x + x_increment
        y = y + y_increment

    return pixels
```

```python
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))
```

```python
pixels = dda_line(x1, y1, x2, y2)

print("\nGenerated Pixels:")
print(pixels)
```

```python
x_pixels = [p[0] for p in pixels]
y_pixels = [p[1] for p in pixels]

plt.figure(figsize=(8, 6))

# Ideal mathematical line
plt.plot(
    [x1, x2],
    [y1, y2],
    label="Ideal Mathematical Line"
)

# Generated pixels
plt.scatter(
    x_pixels,
    y_pixels,
    s=100,
    marker="s",
    color="red",
    label="DDA Pixels"
)

# End points
plt.scatter(
    [x1, x2],
    [y1, y2],
    s=150,
    marker="o",
    color="green",
    label="End Points"
)

# Pixel labels
for x, y in pixels:
    plt.text(
        x + 0.1,
        y + 0.1,
        f"({x},{y})",
        fontsize=9
    )

plt.xticks(
    range(min(x1, x2) - 1, max(x1, x2) + 2)
)

plt.yticks(
    range(min(y1, y2) - 1, max(y1, y2) + 2)
)

plt.grid(True)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.title("DDA Line Drawing Algorithm")

plt.legend()

plt.show()
```
